```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# Dataset

# Read Kaggle dataset
df = pd.read_csv("StudentsPerformance.csv")
df
```

```python
# Task1: Relational plot
# Create relational plot
sns.relplot(data=df,x='math score',y='reading score',hue='gender')

plt.title("Relationship Between Math and Reading Scores")
plt.show()
```

```python
# Task2: Line plot as scatter & facet

sns.lineplot(
    data=df,
    x='math score',
    y='reading score',
    hue='gender'
)

plt.title("Math Score vs Reading Score by Gender")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()
```

```python
# Task3: Distribution Plot
sns.histplot(data=df, x='math score', bins=10)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Frequency")

plt.show()

sns.kdeplot(data=df, x='math score')

plt.title("KDE Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Density")

plt.show()

sns.rugplot(data=df, x='math score')

plt.title("Rug Plot of Math Scores")
plt.xlabel("Math Score")

plt.show()


sns.histplot(
    data=df,
    x='math score',
    bins=10,
    kde=True
)

plt.title("Math Score Distribution with Histogram and KDE")
plt.xlabel("Math Score")
plt.ylabel("Frequency")

plt.show()
```

```python
#TAsk4: Bivariate distribution plot
sns.histplot(
    data=df,
    x='math score',
    y='reading score',
    bins=10
)

plt.title("Bivariate Histogram: Math vs Reading Scores")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()

sns.kdeplot(
    data=df,
    x='math score',
    y='reading score'
)

plt.title("Bivariate KDE: Math vs Reading Scores")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()
```

```python
#Task5: matrix plot

sns.pairplot(
    data=df,
    vars=['math score', 'reading score', 'writing score'],
    hue='gender'
)

plt.show()


correlation_matrix = df[
    ['math score', 'reading score', 'writing score']
].corr()

print(correlation_matrix)

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap of Student Scores")
plt.show()
```

```python
# TAsk6:Categorical plot
sns.barplot(data=df, x='gender', y='math score')
plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")
plt.show()

sns.boxplot(data=df, x='gender', y='math score')
plt.title("Math Score Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Math Score")
plt.show()

sns.violinplot(data=df, x='gender', y='math score')
plt.title("Math Score Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Math Score")
plt.show()

sns.countplot(data=df, x='gender')
plt.title("Number of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.show()
```

```python
# TAsk7: rEgression plot
sns.regplot(
    data=df,
    x='math score',
    y='reading score'
)

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()

sns.lmplot(
    data=df,
    x='math score',
    y='reading score',
    hue='gender'
)

plt.title("Math vs Reading Scores by Gender")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()
```

```python
# Task8: Multi plot and figure level plot
g = sns.FacetGrid(
    data=df,
    col='gender'
)

g.map(
    sns.scatterplot,
    'math score',
    'reading score'
)

g.set_axis_labels("Math Score", "Reading Score")
g.set_titles("Gender: {col_name}")

plt.show()


sns.relplot(
    data=df,
    x='math score',
    y='reading score',
    col='gender',
    kind='scatter'
)

plt.show()


sns.catplot(
    data=df,
    x='lunch',
    y='math score',
    col='gender',
    kind='box'
)

plt.show()

sns.displot(
    data=df,
    x='math score',
    col='gender',
    kind='hist'
)

plt.show()

```
