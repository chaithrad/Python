```python
# Task1:

sales = [1200, 450,980,1500,3000]

# write to file
file = open("sales_data_CommaSep.txt", "w")
for sale in range(len(sales)):
    file.write(f"{sales[sale]}")
    if sale < len(sales) - 1:
        file.write(",")
file.close()

# read from file
file = open("sales_data_CommaSep.txt", "r")
sales_data = file.read()
print("Sales data read from file:")
print(sales_data)
file.close()

```

```python
# Task2:
#1. using .read()
file = open("sales_data_CommaSep.txt", "r")
sales_data = file.read()
print("Sales data read from file using .read():")
print(sales_data)
file.close()

#2. using .readline()
file = open("sales_data_CommaSep.txt", "r")  
sales_data = file.readline()
print("First line of sales data read from file using .readline():")
print(sales_data)
file.close()

#3. using .readlines()
file = open("sales_data_CommaSep.txt", "r")
sales_data = file.readlines()   
print("All lines of sales data read from file using .readlines():")
print(sales_data)
file.close()
```
