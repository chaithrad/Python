```python
# factorial fucntion
def factorial(n):
    if n < 0:
        print("Error:n is negative number.")
        return 
        
    if n == 0 or n == 1:
        return 1
    
    return n * factorial(n - 1)
    
```

```python
# tets 1:
print(factorial(5))  
print(factorial(0))
print(factorial(-3)) 
```
