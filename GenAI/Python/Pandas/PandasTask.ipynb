```python
import pandas as pd
import numpy as np
```

```python
# Task1: pandas series basics

# # panads series from list 
marks = [78,85,90,66,72]
marks_series = pd.Series(marks)

print("Series avalues: ",marks_series.values)
print("Series index: ",marks_series.index)
print("Series data type: ",marks_series.dtype)

# acess 1st and last 2 elements
print("1st element: ",marks_series[0])
print(f"Last 2 elements:\n{marks_series[-2:]}")

```

```python
# Task2: Mathematical operations on series

# add 5 marks grace to all students
marks_series = marks_series + 5
print(f"Series after adding grace marks:\n{marks_series}")

#subtract 2 marks from all students
marks_series = marks_series - 2
print(f"Series after subtracting grace marks:\n{marks_series}")

# multiply by marks 1.05
marks_series = marks_series * 1.05
print(f"Series after multiplying by 1.05:\n{marks_series}")

# devide all marks  by 2
marks_series = marks_series / 2
print(f"Series after dividing by 2:\n{marks_series}")

```

```python
# Task3: Python functionality on series
max_marks = max(marks_series)
min_marks = min(marks_series)
sum_marks = sum(marks_series)
mean_marks = np.mean(marks_series)
print(f"Maximum marks: {max_marks}")
print(f"Minimum marks: {min_marks}")
print(f"Sum of marks: {sum_marks}")
print(f"Mean of marks: {mean_marks}")

# apply lambda function to check if all students passed >=70
passed = marks_series.apply(lambda x: x >= 70)
print(marks_series)
print(f"Students who passed:\n{passed}")

# count how many passd
print(f"Number of students passed: {passed.sum()}")
```

```python
# Task4: Create a dataframe
students = {
    "Name": ['Amit','Neha','Rahul','Sneha','Pooja'],
    "Marks": [78,85,90,66,72],
    "Subject": ['Maths','Maths','Science','Science','Maths']
}

# conevrt to df
df_students = pd.DataFrame(students)
print(f"Dataframe:\n{df_students}")
print(f"First 3 rows of dataframe:\n{df_students.head(3)}")
print(f"Last 2 rows of dataframe:\n{df_students.tail(2)}")
print(f"Dataframe shape: {df_students.shape}")
print(f"Dataframe columns: {df_students.columns}")
```

```python
# Task5:Important Dataframe functions
# use info()
print(f"Dataframe Info: {df_students.info()}")
print(f"Dataframe describe:\n{df_students.describe()}")
print(f"Head of dataframe:\n{df_students.head()}")
print(f"Tail of dataframe:\n{df_students.tail()}")

# sorting in descending order by marks
df_sorted = df_students.sort_values(by='Marks',ascending=False)
print(f"Dataframe sorted by marks:\n{df_sorted}")

# rest the index after sorting
df_sorted.reset_index(drop=True,inplace=True)
print(f"Dataframe with reset index:\n{df_sorted}")

```

```python
# Task6: Filtering and conditional selection

# student scoredmored than 75 marks
high_scorers = df_students[df_students['Marks'] > 75]

# student belong to maths subject
maths_students = df_students[df_students['Subject'] == 'Maths']

# student who scored more than avg  
avg_marks = df_students['Marks'].mean()
above_avg_students = df_students[df_students['Marks'] > avg_marks]

# student who failed (marks < 70)
students_failed = df_students[df_students['Marks'] < 70]
    
print(f"Students who scored more than 75 marks:\n{high_scorers}")
print(f"Students who belong to Maths subject:\n{maths_students}")
print(f"Students who scored above average marks:\n{above_avg_students}")
print(f"Students who failed:\n{students_failed}")
```

```python
# Task7: grouping and basic analysis

# find avg marks per sub using groupby
avg_marks_per_subject = df_students.groupby('Subject')['Marks'].mean()

# count number of student per subject
student_count_per_subject = df_students.groupby('Subject')['Name'].count()

# find max marks per subject
max_marks_per_subject = df_students.groupby('Subject')['Marks'].max()

print(f"Average marks per subject:\n{avg_marks_per_subject}")
print(f"Number of students per subject:\n{student_count_per_subject}")
print(f"Maximum marks per subject:\n{max_marks_per_subject}")
```

```python
# Task 8: Pandas plotting simple graphs

# plot bar grapgh of student vs marks 
df_students.set_index('Name')['Marks'].plot(kind='bar',title='Student Marks', xlabel='Student Name', ylabel='Marks')
```

```python

# plot line graph of marks
df_students.set_index('Name')['Marks'].plot(kind='line', title='Marks Over Time', xlabel='Student Name', ylabel='Marks')
```

```python
# plot histogram of marks
df_students['Marks'].plot(kind='hist', bins=5, edgecolor='black', title='Distribution of Marks', xlabel='Marks', ylabel='Frequency')
```

```python
# TAsk 9: Mini use case: sales analysis

sales = { "Day": ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
          "Revenue": [1200, 1500, 900, 2000, 1800]}


# total revenue for the week
df_sales = pd.DataFrame(sales)
total_revenue = df_sales['Revenue'].sum()
print(f"Total revenue for the week: {total_revenue}")

# avg daily revenue
avg_daily_revenue = df_sales['Revenue'].mean()
print(f"Average daily revenue: {avg_daily_revenue}")

# day with highets revenue
day_highest_revenue = df_sales.loc[df_sales['Revenue'].idxmax()]['Day']
print(f"Day with highest revenue: {day_highest_revenue}")


# day where revnue > avg daily revenue
df_sales[df_sales['Revenue'] > avg_daily_revenue]['Day']
print(f"Days with revenue greater than average daily revenue:\n{df_sales[df_sales['Revenue'] > avg_daily_revenue]['Day']}")


# plot revenue vs day  
df_sales.set_index('Day')['Revenue'].plot(kind='bar', title='Sales Analysis', xlabel='Day', ylabel='Revenue')
```
