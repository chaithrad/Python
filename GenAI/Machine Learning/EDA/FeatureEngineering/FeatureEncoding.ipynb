eg: yellow  red  green 
     0       0     1

     1       0     0

     1       1     1          so there is some kind of dependency on the clumns(multicollinearlty)

so inorder to remove this dependency, we try to make co. of cloum = no. of categories-1:  ===> Dropping one dummy column

   yellow  red

     1       1      here the sum of clumn is 1 but no depenecy  between 2 colum this will avoid  multicollinearlty problem
