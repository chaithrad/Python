# types of Impuatation techniques:
-------------------------------------
1. Remove the missing values
2. Mean, median imputation -> numerics 
3. mode impuataion--> categorical
4. Constant or Sentinal value Impuatation
5. Forward fill / backward  fill (time series data)
6. Simple Impuatation--> can use from scikit learn. It is only going to take input as 2d vector. Also we can convert 2d to 1d using ravell(). There are few paranetre use dby simple imputation they are:
     1. stratergy: it decides that what impuatation we can go with  like mean. median , most frequent etc.

```python
df=pd.read_csv('unclean_smartwatch_health_data.csv')
```

```python
df = df.drop(columns = ['User ID'])
```

```python
df.info()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='median')
```

```python
df['Heart Rate (BPM)'] = imputer.fit_transform(df[['Heart Rate (BPM)']]).ravel()
```

```python
df.info()
```

```python
imputer = SimpleImputer(strategy='mean')
```

```python
df['Blood Oxygen Level (%)'] = imputer.fit_transform(df[['Blood Oxygen Level (%)']]).ravel()
```

```python
df.info()
```

```python
df['Activity Level'] 
```

```python
imputer = SimpleImputer(strategy='most_frequent')
```

```python
df['Activity Level'] = imputer.fit_transform(df[['Activity Level']]).ravel()
```

```python
df.info()
```

```python
imputer = SimpleImputer(strategy='constant',fill_value=5)
```

```python
df['Stress Level'] = imputer.fit_transform(df[['Stress Level']]).ravel()
```

```python
df.info()
```
