```python
# to read_tsv() -- >Parameter 1:  sep  meansseparation --> how values are separated from each other in a agiven file. in tsv file values are separated by tab space. so we can use sep='\t' to read tsv file. tab separted value-tsv
# in csv -> separtaed by ,
# tsv-> separated by tab space
dftsv = pd.read_csv('katalog_gempa_v2.tsv', sep='\t')
dftsv
```

```python
### Parameter 2. names parameter --> is going to list of names that to provide as col names, susch that  names=['name','age','sex']  and then we can use header=0 to tell that first row is header row. if we use header=None then it will not consider first row as header row. it will consider first row as data row. so we can use names parameter to provide our own column names.       
dftitanicnames = pd.read_csv('Titanic-Dataset.csv', names=['P_Id','Survived','Pclass','Name','Gender','Age','Sibling or SpouseSp','Parch','Ticket_id','Fare','Cabin','Starting Journey'])   
dftitanicnames
```

```python
# Parameter3:  header --> header parameter makes sure that any of the row name will be shown as the column name only based on indexing of rows
# # use header = 0 for get rid of actual header row and use our own column names. if we use header=None then it will not consider first row as header row. it will consider first row as data row. so we can use names parameter to provide our own column names.
dftitanicHeader = pd.read_csv('Titanic-Dataset.csv', names=['P_Id','Survived','Pclass','Name','Gender','Age','Sibling or SpouseSp','Parch','Ticket_id','Fare','Cabin','Starting Journey'], header=0)   
dftitanicHeader
```

```python
dftitanicHead = pd.read_csv('Titanic-Dataset.csv', header=4)   
dftitanicHead
```

```python
# Parameter 4: index_col--> help u sto keep one of  our columas index column. so we can use index_col parameter to keep one of our columns as index column. we can use index_col=0 to keep first column as index column. we can use index_col=1 to keep second column as index column. we can use index_col='column_name' to keep any of our column as index column.
dftitanicIdx = pd.read_csv('Titanic-Dataset.csv', index_col='PassengerId')
dftitanicIdx
```

```python
# Parameter 5: usecols --> usecols parameter is used to read only specific columns from a csv file. we can use usecols parameter to read only specific columns from a csv file. we can use usecols=['column_name1','column_name2'] to read only specific columns from a csv file. we can use usecols=[0,1] to read only specific columns from a csv file. we can use usecols=lambda x: x in ['column_name1','column_name2'] to read only specific columns from a csv file.
# use cols--> what columns we should use to show out of all columns we are having in the dataframes. 
# Uses alist to mentionall of teh column names which are required.
dftitanicUsecols = pd.read_csv('student_depression_dataset.csv', usecols=['id',	'Gender',	'Age',	'City',	'Profession',	'Academic Pressure',	'CGPA',	'Study Satisfaction',	'Sleep Duration',	'Dietary Habits',	'Degree',	'Have you ever had suicidal thoughts ?',	'Depression'])
dftitanicUsecols
```

```python
# Parameter 6:sqeeze --> squeeze parameter is used to convert a dataframe into a series. we can use squeeze parameter to convert a dataframe into a series. we can use squeeze=True to convert a dataframe into a series. we can use squeeze=False to keep the dataframe as it is.
#squeze--> dataframe--> series---> squeeze=True--> combination of usecols and squeeze --> True
# can use this squeeze() as squeeze=True is usedin older veriosn pf pandas. so we can use squeeze() to convert a dataframe into a series.
pd.read_csv('Titanic-Dataset.csv',usecols=['PassengerId']).squeeze()
```

```python
# Parameter 7:nrows --> rows parameter is used to read only specific number of rows from a csv file. we can use nrows parameter to read only specific number of rows from a csv file. we can use nrows=5 to read only first 5 rows from a csv file. we can use nrows=10 to read only first 10 rows from a csv file.
## helps us with a fixed number of rows out of the complte dataset.

pd.read_csv('student_depression_dataset.csv',nrows=5)
```

```python
# Encoding issue when working like numerical data, we can use encoding parameter to read the csv file. we can use encoding='utf-8' to read the csv file. we can use encoding='latin-1' to read the csv file. we can use encoding='ISO-8859-1' to read the csv file. we can use encoding='cp1252' to read the csv file. we can use encoding='utf-16' to read the csv file. we can use encoding='utf-32' to read the csv file. we can use encoding='ascii' to read the csv file. we can use encoding='unicode_escape' to read the csv file. we can use encoding='raw_unicode_escape' to read the csv file. we can use encoding='backslashreplace' to read the csv file. we can use encoding='xmlcharrefreplace' to read the csv file.
# so manually keep the encoding -->utf-8--> read the content from the datset
# Parameter 8: encoding parameter make sure we are reading in th uft-8 format---> encoding ='latin-1'
# pd.read_csv('Titanic-Dataset.csv', encoding='latin-1')
pd.read_csv('Titanic-Dataset.csv', encoding='utf-8')
```

```python
# #Parameter 9 : error_bad_lines parameter is used to skip specific number of bad lines from a csv file. we can use error_bad_lines parameter to skip specific number of bad lines from a csv file. we can use error_bad_lines=False to skip specific number of bad lines from a csv file. we can use error_bad_lines=True to keep specific number of bad lines from a csv file.
# error_bad_lines--> we can have some rows with 8 values wjile other will have 9 values so these rows with 8 values are required to be removed --> error_bad_line = False
pd.read_csv('Titanic-Dataset.csv', error_bad_lines=False)
```

```python
# Parameter 10 : skiprows --> skiprows parameter is used to skip specific number of rows from a csv file. we can use skiprows parameter to skip specific number of rows from a csv file. we can use skiprows=5 to skip first 5 rows from a csv file. we can use skiprows=10 to skip first 10 rows from a csv file. we can use skiprows=[0,1,2] to skip first 3 rows from a csv file. we can use skiprows=lambda x: x in [0,1,2] to skip first 3 rows from a csv file.
pd.read_csv('Titanic-Dataset.csv', skiprows=5)
```

```python
# parameter 13: converters --> converters parameter is used to convert specific columns from a csv file. we can use converters parameter to convert specific columns from a csv file. we can use converters={'column_name': function} to convert specific columns from a csv file. we can use converters={'column_name1': function1, 'column_name2': function2} to convert specific columns from a csv file.

def station(value):
    if value =='Q':
        return 'Queentown'
    else:
        return value
```

```python
pd.read_csv('Titanic-Dataset.csv', converters={'Embarked': station})
```

```python
#  parameter 14:na_values --> na_values parameter is used to replace specific values with NaN in a csv file. we can use na_values parameter to replace specific values with NaN in a csv file. we can use na_values=['value1','value2'] to replace specific values with NaN in a csv file. we can use na_values={'column_name': ['value1','value2']} to replace specific values with NaN in a specific column of a csv file. we can use na_values={'column_name1': ['value1','value2'], 'column_name2': ['value3','value4']} to replace specific values with NaN in multiple columns of a csv file.
# this  can help us out with mentioning what values to beconsidered as  amissing value
## na_values = '\',':','No'] etc.
pd.read_csv('Titanic-Dataset.csv', na_values=['NA', 'NaN', 'nan'])
```
